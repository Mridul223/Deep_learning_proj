```python
import numpy as np 
from tensorflow.keras.models import Sequential 
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
Sentense=[
    # Positive sentences
    "I love this movie",
    "This movie is amazing",
    "This movie is terrible",
    "I really dislike this film",
    "I enjoyed this movie",

    # Negative sentences
    "I hate this movie",
    "I really like this film",
    "This is a great movie",
    "This is a bad movie",
    "I did not enjoy this movie"

]
labels=np.array([1,  # I love this movie -> Positive
    1,  # This movie is amazing -> Positive
    0,  # I really like this film -> Positive
    0,  # This is a great movie -> Positive
    1,  # I enjoyed this movie -> Positive

    0,  # I hate this movie -> Negative
    1,  # This movie is terrible -> Negative
    1,  # I really dislike this film -> Negative
    0,  # This is a bad movie -> Negative
    0   ])
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts(Sentense)
```

```python
tokenizer.word_index
```

```python
sent=tokenizer.texts_to_sequences(Sentense)
```

```python
sent
```

```python
X=pad_sequences(
    sent,
    maxlen=6,
    padding="post"
)
```

```python
X
```

```python
vocab_size = len(tokenizer.word_index) + 1
```

```python
vocab_size
```

```python
model=Sequential()
```

```python
model.add(
    Embedding(
        input_dim=vocab_size,
        output_dim=42
    )
)
```

```python
model.add(
    SimpleRNN(
        32,
        activation="tanh",
    )
)
```

```python
model.add(
    Dense(
        1,activation="sigmoid",
        
    )
)
```

```python
model.build(input_shape=(None,6))
```

```python
model.summary()
```

```python
model.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(
    X,
    labels,
    epochs=100,
    verbose=1
)
```

```python
new=["I don't like this"]
```

```python
new_s=tokenizer.texts_to_sequences(new)
```

```python
new_in=pad_sequences(
    new_s,
    maxlen=6,
    padding="post"
)
```

```python
prediction=model.predict(new_in)
```

```python
probability=prediction[0][0]
```

```python
probability
```

```python
if probability >= 0.5:

    print("Sentiment: Positive")

else:


    print("Sentiment: Negative")
```
